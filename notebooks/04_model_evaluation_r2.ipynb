# Model Evaluation — R² and the Meaning of “Good”

This notebook preserves the course section where the multiple-linear-regression model was evaluated with **R²**.

The key goal is to avoid the common mistake: **R² = 0.93 does not mean the model is “93% correct.”**


## 1. 📘 Why evaluation exists

A model is not useful simply because `.fit()` completed without an error.

We need to ask:
- How well does it generalize to unseen data?
- How large are its prediction errors?
- Is it better than a simple baseline?
- Is the metric appropriate for the business problem?
- Is performance stable across relevant subgroups/time periods?

That is why the course held out test data.


## 2. 🧠 Actual vs predicted

On the test set we have:

- `y_test` = actual target values we intentionally held back
- `y_pred` = predictions generated from `X_test`

Evaluation compares these two arrays.


````python
# Typical pattern:
# y_pred = regressor.predict(X_test)

# y_test  -> truth
# y_pred  -> model's estimate
````

## 3. 🔢 R² intuition

A useful form is:

\[
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
\]

Where:

\[
SS_{res}=\sum(y_i-\hat{y}_i)^2
\]

is the model's squared prediction error, and

\[
SS_{tot}=\sum(y_i-\bar{y})^2
\]

is the total variation around the mean target.

Mental model:

> R² compares the fitted model with a naive baseline that predicts the mean of Y.


## 4. 🧠 Interpreting R²

Typical intuition:
- `R² = 1` → perfect predictions on that evaluated data
- `R² = 0` → no improvement over predicting the mean target
- `R² < 0` → worse than the mean baseline
- values closer to 1 often indicate stronger explanatory/predictive fit

But context matters enormously.


## 5. ⚠️ Course correction

The course reported an R² around `0.93` and correctly said it is **not exactly a percentage**.

Do not translate it into:
> “the model is 93% correct.”

And do not adopt:
> “R² should always be in the upper 90s.”

A useful R² depends on:
- noise in the problem
- domain
- target
- sampling
- data leakage risk
- business decision being supported
- comparison with baselines and alternative models

## 8. 🧠 Generalization

The purpose of test data is not bureaucracy.

It asks:

> Did the model learn a relationship that works beyond the exact rows it already saw?

This is **generalization**.

A model can fit training data extremely well and still fail on new data.


## 9. ⚠️ Data leakage

A suspiciously excellent score can be caused by leakage.

Examples:
- future information accidentally included in features
- target-derived columns
- preprocessing performed on the full dataset before splitting
- duplicate entities appearing in both train and test
- temporal problems split randomly instead of chronologically

An FDE/architect should ask **why the metric is high**, not only celebrate it.


## 10. 🏗️ Data-engineering connection

Model evaluation depends on data engineering:
- correct grain
- stable keys
- no duplicates/leakage
- reproducible feature generation
- time-aware data extraction
- versioned datasets
- lineage

A sophisticated model built on leaking or inconsistent data is still a bad system.


## 11. 💊 Pharma lens

For pharma/commercial synthetic exercises, do not evaluate only a global metric.

Eventually ask:
- does error differ by product?
- territory?
- time period?
- account segment?
- low-volume vs high-volume cases?

A single global R² can hide weak performance in important segments.


## 12. 🌱 PlantMind lens

For predictive-maintenance style problems:
- overall R² may be useful for a continuous health score
- but safety-critical regions may matter more
- large under-predictions of degradation can be more costly than small average errors

Evaluation must connect to the operational consequence.
