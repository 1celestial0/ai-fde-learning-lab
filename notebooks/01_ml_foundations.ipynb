# Machine Learning Foundations — The Map Before the Algorithms

This notebook preserves the foundation taught before linear regression: data types, supervised/unsupervised/reinforcement learning, regression vs classification, and the regression family.

The goal is to build a **mental map**, not memorize isolated definitions.


## 1. 📘 Numerical vs Categorical Data

### Numerical data
Represents quantities where arithmetic usually has meaning.

Examples:
- age
- salary
- rainfall
- height
- number of transactions

Numerical can be:
- **continuous** — temperature, weight, revenue
- **discrete** — number of calls, patients, failures

### Categorical data
Represents groups/labels.

Examples:
- color
- dog breed
- state
- product type

Categorical can be:
- **nominal** — no natural order, such as state or color
- **ordinal** — ordered categories, such as low/medium/high

### Important correction
Do not use “number = numerical” as the rule. A ZIP code may contain digits but is categorical because arithmetic on it is meaningless.


## 2. 🧠 Feature vs Target

In supervised ML, think:

**X = input / feature / predictor**

**Y = target / output / response**

Example:

`Rainfall (X) → Crop Yield (Y)`

An “independent variable” does not mean nothing can influence it. In ML, the practical meaning is that it is being used as an input to predict the target.

## 4. 📘 Supervised Learning

Training data contains both:
- known inputs `X`
- known targets `Y`

Mental model:

```text
Known X + Known Y
        ↓
Learning algorithm
        ↓
Learned model
        ↓
New X
        ↓
Predicted Y
```

The model learns from historical examples where the answer was available during training.


## 5. 📘 Unsupervised Learning

There is no known target label for every training example.

Typical goal:
- discover groups
- discover structure
- reduce dimensions
- find patterns

Course shorthand “we do not have the answers, we group things” is useful as a first intuition, but unsupervised learning is broader than clustering.


## 6. 📘 Reinforcement Learning

A more precise mental model than “positive/negative feedback”:

```text
Agent ──action──> Environment
  ↑                  │
  └────reward/state──┘
```

The agent learns a policy that tries to maximize long-term cumulative reward.

Examples:
- game-playing
- robotics
- control systems
- sequential decision-making


## 7. 🧠 Regression vs Classification

### Regression
Predict a continuous numerical value.

Examples:
- house price
- future sales
- crop yield
- temperature
- pipeline execution time

Quick question: **How much?**

### Classification
Predict a class/category.

Examples:
- fraud / not fraud
- churn / no churn
- disease category
- spam / not spam

Quick question: **Which class?**


## 8. 📘 Regression family introduced in the course

### Simple Linear Regression
One input feature:

\[
\hat{Y}=mX+c
\]

### Multiple Linear Regression
Several input features:

\[
\hat{Y}=m_1X_1+m_2X_2+\cdots+m_nX_n+c
\]

### Polynomial Regression
Adds transformed powers of X, such as:

\[
\hat{Y}=b_0+b_1X+b_2X^2+b_3X^3
\]

The curve can be nonlinear in X while the model remains linear in its coefficients.


## 9. 🧠 Real-world linear-regression applications from the course

Examples discussed:

## 10. 🏗️ Data-engineering connection

A machine-learning table can be read like a data-engineering table:

```text
row              = one observation
feature column   = input signal
target column    = value to predict
training dataset = historical data
```

Your existing SQL/data-pipeline thinking is directly reusable. ML adds the model-training/evaluation layer on top of data quality, feature construction, and reproducible processing.
